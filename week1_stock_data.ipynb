```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tickers = ["AAPL", "MSFT", "GOOGL", "AMZN"]
data = yf.download(tickers, start= "2020-01-01", end = "2025-01-01")
data.head()
```

```python
prices = data["Close"]
prices.head()
```

```python
prices.plot(figsize=(12,6), title = "Stock Prices 2020-2025")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend()
plt.show()
```

```python
#daily returns: how much a stock went up or down on a given day
returns = prices.pct_change()
returns.head()
```

```python
returns.hist(bins=50,figsize=(12,8))
plt.suptitle("Distribution of Daily Returns")
plt.show()
```

```python
returns.describe()
#Higher std means more volatile and riskier
```
